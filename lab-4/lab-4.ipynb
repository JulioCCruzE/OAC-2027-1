## 1. Secuenciador Básico

Para el diseño de la unidad de control de una computadora se requieren máquinas de estados que sean capaces de ejecutar algoritmos maś complejos. Haciendo modificaciones y agregando componentes al direccionamiento implícito se puede obtener el **secuenciador básico** que es el encargado de determinar la dirección de la siguiente microinstrucción.

El secuenciador básico debe presentar una dirección a la memoria ROM para poder leer y ejecutar una microinstrucción. La lógica interna del secuenciador determina la fuente específica de la dirección que se usara para leer la memoria ROM, en total son cuatro fuentes:


1. mPC
2. Liga
3. Registro de Transformación
4. Registro de Interrupción

La elección de la fuente se basa en los bits de información que el secuenciador recibe de la microinstrucción presente. A estos bits de información los llamaremos **MI** y pueden ser 4 u 8 dependiendo de la arquitectura.



### Secuenciador básico de 4 MI

Las cuatro micro instrucciones de este tipo de secuenciador son las siguientes:

| Código | Micro Instrucción | Fuente |
|:---:|:---|:---|
| 00 | Paso Contiguo | mPC |
| 01 | Salto condicional | mPC o Liga |
| 10 | Salto de Transformación | Registro de Transformación |
| 11 | Salto de Interrupción | mPC o Registro de Interrupción |


### Secuenciador básico de 8 MI

Las ocho micro instrucciones de este tipo de secuenciador son las siguientes:

| Código | Micro Instrucción | Fuente |
|:---:|:---|:---|
| 000 | Paso Contiguo | mPC |
| 001 | Salto condicional con 0 | mPC o Liga |
| 010 | Salto condicional con 1 | mPC o Liga |
| 011 | Salto de Transformación | Registro de Transformación |
| 100 | Salto de Interrupción con 0 | mPC o Registro de Interrupción |
| 101 | Salto de Interrupción con 1 | mPC o Registro de Interrupción |
| 110 | Salto Forzado | Liga |
| 111 | Wait | Liga |
